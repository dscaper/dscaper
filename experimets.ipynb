### Relative event positioning

```python
import dscaper

dsc = dscaper.Dscaper()

timeline_name = "relative_events_example_2"
library="street_library"
```

```python
timeline_metadata = dscaper.dscaper.DscaperTimeline(name=timeline_name, description="Test timeline")
dsc.create_timeline(timeline_metadata)

event_metadata = dscaper.dscaper.DscaperEvent(library=library, event_time=['const','2.0'], label=['const','car'], source_file=['const','horn_01.wav'])
resp = dsc.add_event(timeline_name, event_metadata)
event_id_1 = resp.content["id"]
print(f"Added event with ID: {event_id_1}")
```

```python

event_metadata = dscaper.dscaper.DscaperEvent(library=library, event_time=['const','4.0'], label=['const','car'], source_file=['const','horn_02.wav'], preceding_event=event_id_1)
resp = dsc.add_event(timeline_name, event_metadata)
event_id_2 = resp.content["id"]
print(f"Added event with ID: {event_id_2}")
print(resp)
```

```python
event_metadata = dscaper.dscaper.DscaperEvent(library=library, label=['const','car'], source_file=['const','horn_02.wav'], preceding_event=event_id_2)
resp = dsc.add_event(timeline_name, event_metadata)
event_id_3 = resp.content["id"]
print(f"Added event with ID: {event_id_3}")
print(resp)
```

```python
event_metadata = dscaper.dscaper.DscaperEvent(library=library, label=['const','car'], source_file=['const','horn_01.wav'], preceding_event=event_id_3, event_time=['const','2.0'])
resp = dsc.add_event(timeline_name, event_metadata)
event_id_4 = resp.content["id"]
print(f"Added event with ID: {event_id_4}")
print(resp)
```

```python
generate_metadata = dscaper.dscaper.DscaperGenerate()
result = dsc.generate_timeline(timeline_name, generate_metadata)
id = result.content["id"]
print(f"Generated soundscape ID: {id}")
```

### Label metadata demo

```python
import dscaper
import json

dsc = dscaper.Dscaper()
```

```python
#  set label metadata
label_metadata_sandbox = json.dumps({"custom_field": "custom_value"})
label_metadata = dscaper.dscaper.DscaperLabel(
    library="demo_metadata",
    label="sink",
    description="Sounds of water sinking in sink",
    sandbox=label_metadata_sandbox
)

dsc.store_label_metadata(label_metadata)
```

```python
label_metadata = dsc.get_label_metadata(library="demo_metadata", label="sink").content
print(f"Metadata: {label_metadata}\n")
print("Audio files metadata:")
for audio in label_metadata["audios"]:
    print(audio)
```

```python
import dscaper
import os

path_to_audio = os.path.join('tests','data','audio')

soundscape_duration = 10.0
seed = 123
sc = dscaper.Scaper(soundscape_duration, None, None)
sc.ref_db = -20
```

```python
sc.add_background(label=('const', 'park'),
                  source_file=('choose', []),
                  source_time=('const', 0))
sc.add_event(label=('const', 'siren'),
             source_file=('choose', []),
             source_time=('const', 0),
             event_time=('uniform', 0, 9),
             event_duration=('truncnorm', 3, 1, 0.5, 5),
             snr=('normal', 10, 3),
             pitch_shift=('uniform', -2, 2),
             time_stretch=('uniform', 0.8, 1.2))
for _ in range(2):
    sc.add_event(label=('choose', ['car_horn', 'siren']),
                 source_file=('choose', []),
                 source_time=('const', 0),
                 event_time=('uniform', 0, 9),
                 event_duration=('truncnorm', 3, 1, 0.5, 5),
                 snr=('normal', 10, 3),
                 pitch_shift=None,
                 time_stretch=None)
for _ in range(2):
    sc.add_event(label=('const', 'human_voice'),
                 source_file=('choose', []),
                 source_time=('const', 0),
                 event_time=('uniform', 0, 9),
                 event_duration=('truncnorm', 3, 1, 0.5, 5),
                 snr=('normal', 10, 3),
                 pitch_shift=None,
                 time_stretch=None)
               
```

```python
audiofile = 'soundscape.wav'
jamsfile = 'soundscape.jams'
txtfile = 'soundscape.txt'
sc.generate(audiofile, jamsfile,
            allow_repeated_label=True,
            allow_repeated_source=True,
            reverb=0.1,
            disable_sox_warnings=True,
            no_audio=False,
            txt_path=txtfile)

```
